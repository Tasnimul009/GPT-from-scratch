```python
with open('input.txt', 'r', encoding='utf-8') as f:
  text = f.read()
```

```python
print("length of dataset in chars: ", len(text))
```

```python
# the first 1000 characters:
print(text[:1000])
```

```python
chars = sorted(list(set(text)))
#print(chars[:100])
vocab_size = len(chars)

print(''.join(chars))
```

```python
# tokenization
stoi = {ch:i for i,ch in enumerate(chars)}
itos = {i:ch for i,ch in enumerate(chars)}

encode = lambda s:[stoi[c] for c in s]
decode = lambda i: ''.join([itos[c] for c in i])

print(encode('Hello Drax'))
print(decode(encode('Hello Drax')))
```

```python
# use pytorch to encode data and store
import torch
data = torch.tensor(encode(text), dtype=torch.long)
print(data.shape, data.dtype)
print(data[:10])
```

```python
n = int(0.9 * len(data))
train_data = data[:n] #(0 -> n-1)
val_data = data[n:] # (n -> end)
```

```python
block_size = 8
train_data[:block_size+1 ]
```

```python
x = train_data[:block_size]
print(x)
y = train_data[1:block_size+1]
print(y)
```

```python
for t in range(len(x)):
  context = x[:t+1]
  target = y[t]
  print(f"when target is {context}, target is {target}")
```
